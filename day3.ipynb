```python
narx = int(input("Mahsulot narxi: "))
chegirma = int(input("Chegirma foizi: "))

print(f"Chegirma {narx * (1 - chegirma / 100) - narx}\n"
      f"Yangi narx {narx * (1 - chegirma / 100)}")
```

```python
import math

a = float(input("Tomon A: "))
b = float(input("Tomon B: "))
c = float(input("Tomon C: "))

perimetr = a + b + c
s = perimetr / 2
yuza = math.sqrt(s * (s - a) * (s - b) * (s - c))

print(f"Perimetr: {perimetr}")
print(f"Yuzasi: {yuza}")
```

```python
kredit_summa = int(input("Kredit miqdori: "))
yillik_foiz = int(input("Yillik foiz stavkasi: "))
muddat = int(input("Kredit muddati: "))

jami_yil = (kredit_summa / 100) * yillik_foiz


print(f"Yillik foiz: {jami_yil}\n"
      f"Umumiy qarz miqdori: {jami_yil * muddat + kredit_summa}")
```
