```python
with open("files/day10-1-vazifa.txt", "r") as f:
    data = f.read().splitlines()
    data = {item.split()[0]: int(item.split()[1]) for item in data}

for key, value in data.items():
    if value >= 90:
        with open("files/day10-1-javob.txt", "a") as f:
            f.write(f"{key} {value}\n")

```

```python
with open("files/day10-2-vazifa.txt", "r") as f:
    data = f.read().replace(".", "").split()

words = {}

for item in data:
    if item in words:
        words[item] += 1
    else:
        words[item] = 1

print(f"Sozlar soni: {len(data)}")

print("Takrorlanmas so'zlar:")
for word, count in words.items():
    if count == 1:
        print(f"  {word}")

print(f"Ko'p uchraydigan so'zlar:")
for word, count in words.items():
    if count == max(words.values()):
        print(f"  {word}: {count}")
```

```python
with open("files/day10-3-vazifa.txt", "r") as f:
    data = [int(x) for x in f.read().split()]
    
print(f"Eng katta raqam: {max(data)}")
print(f"Eng kichik raqam: {min(data)}")
print(f"Ortacha raqam: {sum(data) / len(data)}")
print(f"Juft sonlar yigindisi: {sum(x for x in data if x % 2 == 0)}")
```
