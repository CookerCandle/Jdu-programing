```python
def bolish(a: int, b: int)-> float | str:
    """Bo'lish funksiyasi"""
    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Nolga bolish mumkin emas."

print(bolish(a=10, b=0))
```

```python
import math

def sonlar(a:int)-> float:
    """son ildizini hisoblash funksiyasi"""
    if a < 0:
        raise ValueError("Error: Manfiy son ildizini hisoblash mumkin emas.")
    return math.sqrt(a)

print(sonlar(a=4))
```

```python
def bolishList(a:list[int])-> None:
    """ Bo'lish funksiyasi ro'yxat uchun"""
    if not a:
        raise ValueError("Error: Bo'sh ro'yxat.")

    if 0 in a:
        raise ValueError("Error: Ro'yxatda nol mavjud, nolga bo'lish mumkin emas.")

    for i in a:
        print(i / 100)

bolishList(a=[20, 14, 30, 40, 50])
```
