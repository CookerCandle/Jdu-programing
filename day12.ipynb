```python
class Avtomobil:
    def __init__(self, model, yil, tezlik):
        self.model = model
        self.yil = yil
        self.tezlik = tezlik

    def harakatlan(self):
        print(f"Avtomobil: {self.model}, {self.yil} da ishlab chiqarilgan, maksimal tezligi {self.tezlik} km/h")


class Elektromobil(Avtomobil):
    def __init__(self, model, yil, tezlik, quvat):
        super().__init__(model, yil, tezlik)
        self.quvat = quvat

    def harakatlan(self):
        print(f"Avtomobil: {self.model}, {self.yil} da ishlab chiqarilgan, maksimal tezligi {self.tezlik} km/h, Batareya miqdori {self.quvat}")


a1 = Elektromobil("Tesla", 2023, 150, "100 kWh")
a1.harakatlan()
```

```python
class Shakl:
    def yuza():
        pass


class Doira(Shakl):
    def __init__(self, radius):
        self.radius = radius
        self.nomi = "doira"

    def yuza(self):
        return 3.1415926 * (self.radius**2)


class Kvadrat(Shakl):
    def __init__(self, tomon):
        self.tomon = tomon
        self.nomi = "kvadrat"

    def yuza(self):
        return self.tomon**2


figuralar = [Doira(5), Kvadrat(5)]

for item in figuralar:
    print(item.nomi, item.yuza())
```

```python
class BankHisobi:
    def __init__(self):
        self.__balans = 0

    def get_balans(self):
        return self.__balans

    def set_balans(self, qiymat):
        self.__balans = qiymat

    def deposit(self, miqdor):
        if miqdor > 0:
            self.__balans += miqdor
            print(f"Hisobga {miqdor} qoshildi")

    def withdraw(self, miqdor):
        if 0 < miqdor <= self.__balans:
            self.__balans -= miqdor
            print(f"{miqdor} som yechildi. Yangi balans: {self.__balans}")
        else:
            print("Mablag yetarli emas yoki notogri miqdor")


class BiznesHisob(BankHisobi):
    def withdraw(self, miqdor):
        komissiya = miqdor * 0.05
        jami_yechim = miqdor + komissiya

        if 0 < jami_yechim <= self.get_balans():
            yangi_balans = self.get_balans() - jami_yechim
            self.set_balans(yangi_balans)

            print(
                f"{miqdor} som yechildi (komissiya bilan {int(jami_yechim)}). Yangi balans: {int(yangi_balans)}"
            )
        else:
            print("Mablag' yetarli emas (komissiya билан бирга)")


hisob = BiznesHisob()
hisob.deposit(1000)
hisob.withdraw(100)
```
