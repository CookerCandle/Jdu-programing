```python
baho = int(input("Baho kiriting: "))

if 90 <= baho <= 100:
    status = "Alo"
elif 75 <= baho < 90:
    status = "Yaxshi"
elif 50 <= baho < 75:
    status = "Qoniqarli"
elif 0 <= baho < 50:
    status = "Qoniqarsiz"
else:
    status = "xatolik"

print(status)
```
