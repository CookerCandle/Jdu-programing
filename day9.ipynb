```python
def full_name(*full_name: str) -> None:
    name_len = len(full_name)
    if name_len == 0:
        print("ism kiritilmadi")
    elif name_len == 1:
        print(full_name[0], "familiyasi yoq")
    else:
        print(full_name[0], full_name[1])

full_name("Ali", "Valiyev")
```

```python
# k = kvadrat
# t = to'gri burchak

def shape_area(shape: str, a: int, b: int | None = None) -> None:
    if shape in ['k', 't']:
        if shape == 'k':
            print(a * a)
        elif shape == 't':
            if b is None:
                raise ValueError("[t] ga (b) parametiri zarur")
            print(a * b)
    else:
        raise ValueError("shape ga [k] yoki [t] bering")

shape_area('k', a=5)
shape_area('t', a=4, b=5) 
```

```python
def sum_to_n(n: int) -> int:
    if n <= 1:
        return max(0, n)
    
    return n + sum_to_n(n - 1)

print(sum_to_n(5))
```
