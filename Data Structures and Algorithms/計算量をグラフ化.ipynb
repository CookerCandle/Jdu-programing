n,2^n, n! の線形スケール

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma

def plot_linear_separated():
    # nを1から20までの範囲に設定（滑らかな曲線を描くため500分割）
    n = np.linspace(1, 20, 500)

    # 3つの関数を計算
    y_n = n
    y_2n = 2**n
    y_fact = gamma(n + 1)  # n! の連続版（ガンマ関数）

    # 描画設定
    plt.figure(figsize=(10, 6))

    # 線形スケール (plot) で描画し、線を太く設定
    plt.plot(n, y_n, label='$n$', linewidth=3)
    plt.plot(n, y_2n, label='$2^n$', linewidth=3)
    plt.plot(n, y_fact, label='$n!$', linewidth=3, linestyle='--')

    # グラフの体裁
    plt.title("Comparison of $n$, $2^n$, and $n!$ (n=20, Thick Lines, Linear Scale)")
    plt.xlabel("n")
    plt.ylabel("Value")

    # ★ここが最大のポイントです★
    # 線形スケールで3本を「はっきり分ける」ために、Y軸の上限を 10,000 に制限します。
    # こうすることで、n! が急上昇して画面外へ消えた後、2^n が上昇していく様子と、
    # その下を這う n の違いが線形スケール上で明確に確認できます。
    plt.ylim(0, 10000)

    # 凡例がグラフに被らないように外側に配置
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()

    # Colab上でグラフを表示
    plt.show()

# プログラムの実行
plot_linear_separated()
```

n, 2^n, n!をlogスケールで表示

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma

def plot_clearly_separated():
    # nを1から20までの範囲に設定（滑らかにするため200分割）
    n = np.linspace(1, 20, 200)

    # 3つの関数を計算
    y_n = n
    y_2n = 2**n
    y_fact = gamma(n + 1)  # n! の連続版（ガンマ関数）

    # 描画設定
    plt.figure(figsize=(10, 6))

    # 3本をはっきり分けるため、Y軸を対数スケール(semilogy)に設定
    # 線の太さを太め (linewidth=3) に指定
    plt.semilogy(n, y_n, label='$n$', linewidth=3)
    plt.semilogy(n, y_2n, label='$2^n$', linewidth=3)
    plt.semilogy(n, y_fact, label='$n!$', linewidth=3, linestyle='--')

    # グラフの体裁
    plt.title("Comparison of $n$, $2^n$, and $n!$ (n=20, Thick Lines, Log Scale)")
    plt.xlabel("n")
    plt.ylabel("Value (Log Scale)")

    # 凡例がグラフに被らないように外側に配置
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=12)
    plt.grid(True, which="both", linestyle='--', alpha=0.5)
    plt.tight_layout()

    # Colab上でグラフを表示
    plt.show()

# プログラムの実行
plot_clearly_separated()
```
