ソートのアルゴリズムの違いによる処理時間の違い

```python
import random
import time
import sys
import copy
import matplotlib.pyplot as plt

# クイックソート実行時にPythonの再帰深度上限に引っかかるのを防ぐ
sys.setrecursionlimit(200000)

# 1. バブルソート
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

# 2. 挿入ソート
def insertion_sort(arr):
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# 3. クイックソート (In-place実装)
def quick_sort(arr, low=0, high=None):
    if high is None:
        high = len(arr) - 1
    if low < high:
        pi = partition(arr, low, high)
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)
    return arr

def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i = i + 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

# 4. ヒープソート
def heapify(arr, n, i):
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2
    if left < n and arr[i] < arr[left]:
        largest = left
    if right < n and arr[largest] < arr[right]:
        largest = right
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]
        heapify(arr, i, 0)
    return arr

def main():
    # 計測するデータサイズのリスト（100 から 10,000）
    sizes = [100, 500, 1000, 2000,3000, 4000, 5000, 6000, 7000, 8000, 9000, 10000]

    times_bubble = []
    times_insertion = []
    times_quick = []
    times_heap = []

    print("ソートアルゴリズムの処理時間計測を開始します。")
    print("※注意: データサイズが大きくなると、バブルソートと挿入ソートには数十分以上の時間がかかる場合があります。\n")

    for size in sizes:
        print(f"データサイズ: {size:6d} の計測中...")

        # 1〜10000の範囲でランダムなデータを生成
        original_data = [random.randint(1, 10000) for _ in range(size)]

        # バブルソート
        data = copy.deepcopy(original_data)
        start = time.time()
        bubble_sort(data)
        times_bubble.append(time.time() - start)

        # 挿入ソート
        data = copy.deepcopy(original_data)
        start = time.time()
        insertion_sort(data)
        times_insertion.append(time.time() - start)

        # クイックソート
        data = copy.deepcopy(original_data)
        start = time.time()
        quick_sort(data)
        times_quick.append(time.time() - start)

        # ヒープソート
        data = copy.deepcopy(original_data)
        start = time.time()
        heap_sort(data)
        times_heap.append(time.time() - start)

    print("\n計測が完了しました。グラフを描画します。")

    # グラフの描画設定
    # ※環境依存の文字化け（豆腐文字）を防ぐため、ラベルは英語で記述しています
    plt.figure(figsize=(10, 6))
    plt.plot(sizes, times_bubble, label='Bubble Sort', marker='o')
    plt.plot(sizes, times_insertion, label='Insertion Sort', marker='s')
    plt.plot(sizes, times_quick, label='Quick Sort', marker='^')
    plt.plot(sizes, times_heap, label='Heap Sort', marker='d')

    plt.title('Sorting Algorithms Processing Time')
    plt.xlabel('Number of Elements (Data Size)')
    plt.ylabel('Time (seconds)')
    plt.legend()
    plt.grid(True)

    # 差が小さすぎるクイックソート/ヒープソートを見やすくしたい場合は、
    # 以下の行のコメントアウトを外してY軸を対数スケールにしてください。
    # plt.yscale('log')

    plt.tight_layout()
    plt.show()

if __name__ == "__main__":
    main()
```
